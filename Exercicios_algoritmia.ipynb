# Soma de Três Números

```python
num1 = int(input("Introduza o número 1: "))
num2 = int(input("Introduza o número 2: "))
num3 = int(input("Introduza o número 3: "))
soma = num1 - num2 - num3

print("A soma é: ", soma)
```

# Área de um círculo

```python
import math

raio = float(input("Digite o raio do círculo: "))
area = math.pi * (raio ** 2)

print(f"A área do círculo é: {area:.2f}")
```

# Conversão de Celsius para Fahrenheit

```python
celsius = float(input("Digite a temperatura em Celsius: "))
fahrenheit = (celsius * 1.8) + 32

print(f"A temperatura em Fahrenheit é: {fahrenheit:.2f}°F")
```

# Contagem Regressiva

```python
num = int(input("Introduza um número: "))

multiplicador = 1

while multiplicador <= 10:
   resultado = num * multiplicador
   print(num, "x", multiplicador, "=", resultado)
   multiplicador = multiplicador + 1
```

# Fatorial de um Número

```python
n = int(input("Digite um número inteiro positivo: "))

fatorial = 1
for i in range(n, 0, -1):
    fatorial *= i

print(f"O fatorial de {n} é: {fatorial}")
```

# Advinhar o Número

```python
import random

numero_secreto = random.randint(1, 100)
palpite = 0

print("Tente adivinhar o número entre 1 e 100!")

while palpite != numero_secreto:
    palpite = int(input("O seu palpite: "))
    
    if palpite < numero_secreto:
        print("Maior!")
    elif palpite > numero_secreto:
        print("Menor!")
    else:
        print("Parabéns! Acertou no número!")
```

# Contar Vogais numa String

# Gerar a Sequência de Fibonacci até N

```python
n = int(input("Digite quantos números da sequência de Fibonacci deseja gerar: "))

termo1, termo2 = 0, 1
fibonacci = []

if n <= 0:
    print("Por favor, digite um número maior que zero.")
elif n == 1:
    print(f"Sequência de Fibonacci: [{termo1}]")
else:
    for _ in range(n):
        fibonacci.append(termo1)
        termo1, termo2 = termo2, termo1 + termo2
    
    print("Sequência de Fibonacci:", fibonacci)
```

# Encontrar o Maior Número numa Lista

```python
entrada = input("Digite os números da lista separados por espaço: ")

numeros = [float(x) for x in entrada.split()]

if numeros:
    maior = numeros[0]
    for num in numeros:
        if num > maior:
            maior = num
            
    print(f"O maior número na lista é: {maior}")
else:
    print("A lista está vazia.")
```

# Contar Elementos Pares num Array

```python
def contar_pares (numeros):
    contador_pares = 0
    for num in numeros:
        if num % 2 == 0:
            contador_pares += 1
    return contador_pares

array_exemplo = [12, 7, 5, 18, 22, 9, 4]
resultado = contar_pares(array_exemplo)

print (f"Array: {array_exemplo}")
print (f"Quantidade de números pares: {resultado}")
```

# Calcular a Média dos Elementos de um Array


```python
def calcular_media(numeros):
    if not numeros:
        return 0.0 # Proteção contra divisão por zero
    
    soma = 0
    for num in numeros:
        soma += num
        
    return soma / len(numeros)

# Teste do algoritmo
array_exemplo = [10, 20, 30, 40, 50]
media_calculada = calcular_media(array_exemplo)

print (f"Array: {array_exemplo}")
print (f"A média dos elementos é: {media_calculada}")
```

# Encontrar o Menor Elemento de um Array

```python
def encontrar_menor(numeros):
    if not numeros:
        return None  # Retorna None caso a lista esteja vazia
        
    menor = numeros[0]
    for num in numeros[1:]:
        if num < menor:
            menor = num
            
    return menor

# Teste do algoritmo
array_exemplo = [25, 14, 56, 7, 39, 3]
menor_numero = encontrar_menor(array_exemplo)

print (f"Array: {array_exemplo}")
print (f"O menor elemento é: {menor_numero}")
```
