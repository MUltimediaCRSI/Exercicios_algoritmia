# 6. Funções embutidas nas strings

```python
texto = 'PNG é um formato de arquivo de imagem digital que usa compressão sem perdas,' \
    'criado em 1996 para suportar planos de fundo transparentes e milhões de cores.'

lenString = len(texto) # Comprimento de uma string ou coleção de dados
print (lenString)

numero_palavras = len(texto.split()) # Separa as palavras para uma lista e a len conta-as
print (numero_palavras)

findString = texto.find ('f') # índice da primeira ocorrência de um elemento 
print (findString)

countString = texto.count ('arquivo')
print (countString)

replaceString = texto.replace ('arquivo' , 'ficheiro')
print (replaceString)

```

### 6.1 Contagem de palavras de um texto (nº de vezes que aparece cada palavra)

```python
texto = 'PNG é um formato de arquivo de imagem digital que usa compressão sem perdas,' \
    'criado em 1996 para suportar planos de fundo transparentes e milhões de cores.'

palavras = texto.lower().replace ("." , "").split()

frequencia = {}

for palavra in palavras:

    frequencia[palavra] = frequencia.get(palavra, 0) + 1
    # Já existe esta palavra? Se existir, diz-me quantas vezes. Se não exixtir, dá-me o valor 0.

print(frequencia)

# Função que melhora a aparencia do print

for palavra, quantidade in frequencia.items():
    print (f'{palavra} : {quantidade}')

```

```python
from wordcloud import WordCloud, STOPWORDS 
import matplotlib.pyplot as plt

# Importamos a classe Wordcloud para gerar uma nuvem de palavras e stopwords para escolher o número de palavras que quero

# texto = "Na disiciplina de Multimédia vamos aprender algumas coisas interessantes."
texto = 'PNG é um formato de arquivo de imagem digital que usa compressão sem perdas,' \
    'criado em 1996 para suportar planos de fundo transparentes e milhões de cores.'

# Conjunto de palavras removidas do texto
as_minhas_stopwords = set(['Na' , 'de' , 'algumas' , '.'])

# -------------------- Configurar e gerar a nuvem de palavras a partir do texto ---------------

nuvem = WordCloud (
    width = 1000, # px
    height = 600,
    background_color = 'white',
    colormap = 'viridis', # 'plasma' ; 'Blues' ; 'Reds' ; 'Greens' ...
    max_words = 12,
    stopwords = as_minhas_stopwords
).generate(texto)

# ------------ Exibir a imagem através de módulo Matplotlib -----------

plt.figure(figsize=(10, 5)) # define o tamanho
plt.imshow(nuvem, interpolation = 'bilinear') # rederiza a imagem
plt.axis ('off') # oculta o x e y
plt.show() # mostra a figura

nuvem.to_file('nuvem_de_palavras.jpg')
```
